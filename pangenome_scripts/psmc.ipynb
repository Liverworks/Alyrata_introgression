```python
try_model = pd.read_csv("LDs/psmc/Eur_lyr_raw_plot.ssv", sep='\t', header=None)
try_model
```

```python
scatter = sns.lineplot(x=try_model.iloc[:,0], y=try_model.iloc[:,1], hue=try_model.iloc[:,5])
plt.xscale('log')
scatter.set_xlim(left=10000)
```

```python
eur_list = ["AL27", "AL56", "AL85", "AL08", "BOR", "73_3a", "NLA", "ALR", "F1-14"]
esib_list = ["NT12", "NT8", "NT9", "BAM12.1", "BAM12.3"]
wsib = ["TE11", "TE4", "TE8"]
scand = ["S06", "S18", "N7", "I04"]
nam = ["MN47", "TSS"]
```

```python
try_model["pop"] = ""
```

```python
try_model["pop"][try_model.iloc[:,5].isin(eur_list)] = "Central Europe"
try_model["pop"][try_model.iloc[:,5].isin(esib_list)] = "East Siberia"
try_model["pop"][try_model.iloc[:,5].isin(wsib)] = "Central Siberia"
try_model["pop"][try_model.iloc[:,5] == "Kar"] = "Karelia"
try_model["pop"][try_model.iloc[:,5].isin(nam)] = "North America"
try_model["pop"][try_model.iloc[:,5].isin(scand)] = "Scandinavia"
```

```python
try_model
```

```python
scatter = sns.lineplot(x=try_model.iloc[:,0], y=try_model.iloc[:,1], hue=try_model.iloc[:,6])
plt.xscale('log')
scatter.set_xlim(left=10000)
```

```python
try_model
```

```python
samp_list = ["AL27", "AL56", "AL85", "AL08", "BOR", "MN47", "NT12", "NT8", "NT9", "BAM12.1", "BAM12.3", "TE11", "TE4", "TE8","TSS", "Kar","S06", "S18", "I04", "N7", "73_3a", "NLA", "ALR", "F1-14"]
for i in samp_list:
    plot_tab = try_model[try_model.iloc[:,5] == i]
    if i in eur_list:
        scatter = sns.lineplot(x=plot_tab.iloc[:,0], y=plot_tab.iloc[:,1], color="#00b9e3") #, label="Central Europe")
    elif i == "MN47":
        scatter = sns.lineplot(x=plot_tab.iloc[:,0], y=plot_tab.iloc[:,1], color="#00ba38") #, label="North America")
    elif i in esib_list:
        scatter = sns.lineplot(x=plot_tab.iloc[:,0], y=plot_tab.iloc[:,1], color="#93aa00") #, label="East Siberia")
    elif i in wsib:
        scatter = sns.lineplot(x=plot_tab.iloc[:,0], y=plot_tab.iloc[:,1], color="#db72fb") #, label="Central Siberia")
plt.xscale('log')
scatter.set_xlim(left=10000)
plt.savefig("psmc_python_groups.pdf")
#plt.legend(handles=["#00ba38", "#00b9e3", "#db72fb", "#93aa00"])
```
