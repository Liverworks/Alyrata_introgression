```python
pca = pca.rename(columns={1: "newer_name"})
```

```python
pca = pca.merge(metadata, on="newer_name", how="left")
```

```python
fig = plt.figure(figsize = (8,8))

ax = fig.add_subplot(1,1,1) 
sns.scatterplot(x=pca.iloc[:, 2][pca.loc[:,"Subspecies"] != "-"], y=pca.iloc[:, 3][pca.loc[:,"Subspecies"] != "-"], hue=pca.loc[:,"Subspecies"][pca.loc[:,"Subspecies"] != "-"])

ax.set_xlabel('PC1 (43%)')
ax.set_ylabel('PC2 (22%)')
plt.savefig('../Eur_lyrata/Lyrata_subspecies_pca.pdf')
```
